```python
import numpy as np
import matplotlib.pyplot as plt
k=8.99e9
q=1
N=20
x0 = 6.0
y0=7.0 
x = np.linspace( x0 - 6.5 ,x0 + 6.5,N)+1e-6 # so as not to have point at zero
y = np.linspace(y0 - 7.0  , x0+7.0,N)+1e-6
X,Y = np.meshgrid(x,y)

# field for point 1 (top vertex)
dx= X- x0
dy= Y- y0
r2=dx**2+dy**2
# now find field
thE = np.arctan2(dy,dx)
magE = k*q/r2
Ex1 = magE*np.cos(thE)
Ey1 = magE*np.sin(thE)

# field for point 2 (bottom left vertex)
x0_2 = 1.0
y0_2 = 2.0
dx2 = X - x0_2
dy2 = Y - y0_2
r2_2 = dx2**2 + dy2**2
# now find field
thE2 = np.arctan2(dy2,dx2)
magE2 = k*q/r2_2
Ex2 = magE2*np.cos(thE2)
Ey2 = magE2*np.sin(thE2)

# field for point 3 (bottom right vertex)
x0_3 = 11.0
y0_3 = 2.0
dx3 = X - x0_3
dy3 = Y - y0_3
r2_3 = dx3**2 + dy3**2
# now find field
thE3 = np.arctan2(dy3,dx3)
magE3 = k*q/r2_3
Ex3 = magE3*np.cos(thE3)
Ey3 = magE3*np.sin(thE3)

# combine fields
Ex = Ex1 + Ex2 + Ex3
Ey = Ey1 + Ey2 + Ey3
magE = np.sqrt(Ex**2 + Ey**2)

mask = magE > 40 # will have big values near origin

# quiver plot
fig, ax = plt.subplots()
q = ax.quiver(X[mask],Y[mask],Ex[mask],Ey[mask])
ax.set_aspect('equal')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title(f'Electric Field of 3 Point Charges (Triangle)')

plt.show()
```
